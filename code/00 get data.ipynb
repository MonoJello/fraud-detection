```python
import pandas as pd
from scipy.io import arff
import requests
from utils import describe as des
```

```python
url = "https://www.openml.org/data/download/21756045/dataset"
output_file = "../data/dataset.arff"

response = requests.get(url)
response.raise_for_status()

with open(output_file, "wb") as file:
    file.write(response.content)
```

```python
data, metadata = arff.loadarff("../data/dataset.arff")

df = pd.DataFrame(data)

# Convert byte-string columns to regular strings
for column in df.select_dtypes(include="object").columns:
    df[column] = df[column].str.decode("utf-8")
```
